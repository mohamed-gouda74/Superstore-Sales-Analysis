## Exploratory Data Analysis (EDA)

#### Kye Performance Indicators (KPI) Overview

```python
# Calculate core KPIs

total_salse = df['Sales'].sum()

total_profit = df['Profit'].sum()

total_quantity = df['Quantity'].sum()

profit_margin = (total_profit / total_salse)*100

avg_sales_by_order = df.groupby(['Order ID'])['Sales'].sum().mean()
```

```python
# Disply KPIs 

print(f" Total Sales : ${total_salse:,.2f}")

print(f"Total Profit : ${total_profit:,.2f}")

print(f"Total Units Sold : {total_quantity:,} units")

print(f"Avg Sales per order : ${avg_sales_by_order:,.2f} ")

print(f"Profit Margin : {profit_margin:.2f}%")
```

 **Key Business KPIs Summary:**
 - **Total Sales:** $2.30M
 - **Total Profit:** $286.40K
 - **Total Units Sold:** 37,873 units
 - **Average Sales per Order:** $458.61
 - **Profit Margin:** 12.47%
 
 **Insight:** The business generated over $2.3M in sales with a solid 12.47% profit margin. On average, each customer order generates $458.61 across 37,873 units sold.
